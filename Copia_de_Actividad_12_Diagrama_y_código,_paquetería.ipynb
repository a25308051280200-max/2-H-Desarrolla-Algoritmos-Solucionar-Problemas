```python
print("==Sistema de paqueteria==")
peso = float(input("Ingrese el peso del paquete en kg: "))
distancia = float(input("Ingrese la distancia en km: "))
fragil = input("¿El paquete es fragil? (s/n): ")
if peso <= 5:
  costo_base = 100
else:
  costo_base = 200
extras = 0
if distancia <= 100:
  extras += 80
if fragil.lower() == "s": # Assuming 's' for yes based on the prompt
  extras += 50
total = costo_base + extras
print("El costo base: $", costo_base)
print("Los extras: $", extras)
print("El costo total: $", total)
```
