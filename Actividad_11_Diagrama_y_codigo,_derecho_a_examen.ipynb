```python
print("=====DERECHO A EXAMEN ====")
Total_clase = int(input("ingrese el total de clases del semestre: "))
asistencia = int(input("ingrese las asistencias del alumno: "))
if Total_clase == 0:
  print("El total de clases no puede ser cero. Asumiendo 0% de asistencia.")
  Porcentaje = 0
else:
  Porcentaje = (asistencia/Total_clase) * 100
if Porcentaje >= 80:
  print("el alumno puede presentar el examen")
else:
  print("el alumno no puede presentar el examen")
```
